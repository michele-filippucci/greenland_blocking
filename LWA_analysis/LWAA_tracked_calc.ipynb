```python
ds_filt,dic_filt = BT.FilterEvents(ds,dic_unfilt,var_name = "LWAA",
                     pers_min = 5,min_avg_area = 3e5,max_dist=1400)
print("ContourTracking2D function correclty executed")
#saving data
np.save(fn_dic,dic_filt)
ds_filt.to_netcdf(fn_out)
print("Data created")
  
print("--- %s seconds ---" % (time.time() - start_time))
```
