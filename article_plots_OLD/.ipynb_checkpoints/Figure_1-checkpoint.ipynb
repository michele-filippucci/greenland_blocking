### FIRST LET'S CREATE A MASK OVER GREENLAND

```python
# Load the world shapefile from GeoPandas datasets
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

# Filter the dataset to only include Greenland
greenland = world[world['name'] == 'Greenland'].explode(index_parts=False)

# Define the resolution and create a grid of latitudes and longitudes
lat_resolution = 2.5  # Resolution of 2.5 degrees
lon_resolution = 2.5  # Resolution of 2.5 degrees

# Define latitude and longitude ranges (rough bounds around Greenland)
latitudes = np.arange(0, 91, lat_resolution)
longitudes = np.arange(-180, +180, lon_resolution)

# Initialize a mask with zeros
mask = np.zeros((len(latitudes), len(longitudes)))

# Create the mask: 1 inside Greenland, 0 outside
for i, lat in enumerate(latitudes):
    for j, lon in enumerate(longitudes):
        point = Point(lon, lat)
        if any(greenland.contains(point)):
            mask[i, j] = 1

# Convert the mask to an xarray DataArray
mask_xr = xr.DataArray(mask, coords=[latitudes, longitudes], dims=["lat", "lon"])
mask_bool = mask_xr.data == 1

# Convert to an xarray Dataset
ds_mask = xr.Dataset({'mask': mask_xr})

# Set the latitude and longitude attributes for clarity
ds_mask['mask'].attrs['long_name'] = 'Greenland Mask'
ds_mask['lat'].attrs['units'] = 'degrees_north'
ds_mask['lon'].attrs['units'] = 'degrees_east'
```

```python
#LET's MAKE A PLOT TO VISUALIZE THE DOMAIN
plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.set_extent([-80, -10, 50, 90], crs=ccrs.PlateCarree())

# Plot the mask data
plt.contourf(ds_mask['lon'], ds_mask['lat'], ds_mask['mask'], cmap='Blues', transform=ccrs.PlateCarree())

# Add coastlines and gridlines
ax.add_feature(cfeature.COASTLINE)
ax.gridlines(draw_labels=True)

# Add title
plt.title('Greenland Mask (1 = Greenland, 0 = Elsewhere)')

# Show the plot
plt.show()

# Display the xarray dataset structure
print(ds_mask)
```

```python
ds_filt_DAV = xr.open_dataset('ERA5_northem_2.5x2.5_zg_DAV_tracked_daily_1940-2023.nc')
ds_filt_GHA = xr.open_dataset('ERA5_northem_2.5x2.5_zg_GHA_tracked_daily_1940-2023.nc')
#Greenland coordinates
#convert the tracked DAV into a boolean matrix
DAV = xr.where(ds_filt_DAV['DAV_tracked']>0,1,0)
GHA = xr.where(ds_filt_GHA['GHA_tracked']>0,1,0)
times = ds_filt['time']
lons = ds_filt['lon']
lats = ds_filt['lat']
```

```python
def running_mean(data,window):
    new_data = []
    for i in range(len(data)-window):
        new_data.append(np.sum(data[i:i+window])/window)
    #print(len(new_data))
    return new_data
```

```python
count_DAV = 0
count_GHA = 0
count_m = 0
freq_per_year_DAV = []
freq_per_year_GHA = []
months = ['06','07','08']
mask_size = np.sum(mask_xr,axis=(0,1))
tmp = 0
for t,time in tqdm(enumerate(times)):
    DAV_gr = DAV.values[t,mask_bool]
    GHA_gr = GHA.values[t,mask_bool]
    #print(DAV_gr)
    month = str(time.values)[5:7]
    time_diff = t-tmp
    if month in months:
        count_m += mask_size
        count_DAV += np.sum(DAV_gr,axis=(0))
        count_GHA += np.sum(GHA_gr,axis=(0))
    if time_diff >= 365:
        #print(count_m)
        #print(count)
        tmp = t
        freq_per_year_DAV.append(count_DAV/count_m)
        freq_per_year_GHA.append(count_GHA/count_m)
        count_DAV = 0
        count_GHA = 0
        count_m = 0
```

```python
def fit_with_bootstrap(freq_per_year,years,N_it):
    pends = []
    count = 0
    for i in range(N_it):
        tmp = np.random.choice(freq_per_year, size=len(freq_per_year),replace=False)
        coefficients = np.polyfit(years, tmp, 1)
        pend = coefficients[0]
        pends.append(pend)
    coefficients = np.polyfit(years, freq_per_year, 1)
    pend_exp = coefficients[0]
    for pend in pends:
        if abs(pend) > abs(pend_exp):
            count += 1
    p_value = count/N_it
    return pend_exp,p_value
```

```python
window = 10
freq_per_year_run_avg_DAV = running_mean(freq_per_year_DAV,window)
freq_per_year_run_avg_GHA = running_mean(freq_per_year_GHA,window)
years = np.arange(1940,2024,1)
years_run_avg = years[int(window/2):int(-window/2)]

# Fit linear regression model with errors
pend,p_value = fit_with_bootstrap(freq_per_year_DAV,years,2000)
print('DAV')
print(pend,p_value)
pend,p_value = fit_with_bootstrap(freq_per_year_GHA,years,2000)
print('GHA')
print(pend,p_value)

fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(8, 6))

axes[0].plot(years,freq_per_year_DAV,color='black',label='freq. per year')
axes[0].plot(years_run_avg,freq_per_year_run_avg_DAV,color='g',label='freq. 10 years running mean')
axes[0].set_title('DAV')
axes[0].legend()

axes[1].plot(years,freq_per_year_GHA,color='black',label='freq. per year')
axes[1].plot(years_run_avg,freq_per_year_run_avg_GHA,color='g',label='freq. 10 years running mean')
axes[1].set_title('GHA')
axes[1].legend()

```
