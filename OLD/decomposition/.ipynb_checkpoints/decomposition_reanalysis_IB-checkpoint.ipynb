In this notebook we decompose the difference between historial and present data in the geopotential height field in the Northern Atlantic. The decomposition is aimed at discrimating the influence of atmospheric blocking and zonal regime on the observed anomaly.
The decomposition  is the following: <br> <br>
$ z-\hat{z} = [ fz_b + (1-f)z_z ] - [ \hat{f}\hat{z_b} + (1-f)\hat{z_z} ] = (f-\hat{f})(z_b-z_z) + \hat{f}[(z_b-\hat{z_b})-(z_z-\hat{z_z})] + z_z-\hat{z_z} $ <br> <br>
where z_b is the geopotential when blocked. z_z is the zonal geopotential. ^ stays for reanalysis and no sign is for simulated. The various term of the third member are:
- frequency bias
- difference in blocking pattern
- difference in mean state

We are going to plot the various parts of the third memeber for different points.t points.

```python
#IMPORT DATA
reanalysis_hist = "/home/michele/DATA/ERA5/ERA5_northem_2.5x2.5_Z500_daily_1960-1990_jja_DAV_tracked_pers4_ml28.nc"
reanalysis_pres = "/home/michele/DATA/ERA5/ERA5_northem_2.5x2.5_Z500_daily_1990-2020_jja_DAV_tracked_pers4_ml28.nc"
output_prefix = "/home/michele/prog/summer_greenland_block_prel/decomposition"
ds_hist = xr.load_dataset(reanalysis_hist)
ds_pres = xr.load_dataset(reanalysis_pres)
lon=ds_hist["lon"]
lat=ds_hist["lat"]
```

```python
#coordinates of the point where we look for blocking
region = "Greenland" #you can use Europe, Atlantic_Ocean or Greenland
#plon,plat = BT.GeoAreaCoordinates(region) 
plon = -35#7.5   #Davini et D'Andrea 2016 Euro Atlantic box
plat = 70#52.5
print(plon,plat)
#grid dimension
grid = 2.5
g = 9.80665  
l_lon = 60#33.75 #edge of the box along longitude (semilato)
l_lat = 15#22.5 #edge of the box along latitude
```

```python
#DATA COMPUTATION
if computation:

  #HISTORICAL
  version = "hist"
  N_h = 0
  N_z = 0
  zg = np.zeros(np.shape(ds_hist["zg"].values[0,:,:]))
  zg_h = np.zeros(np.shape(ds_hist["zg"].values[0,:,:]))
  zg_z = np.zeros(np.shape(ds_hist["zg"].values[0,:,:]))
  TM90 = ds_hist["DAV"]
  TM90 = TM90[:,:,:]
  TM90 = xr.where(TM90.loc[:,:,:] > 0,1,0) #don't know why but it was just the same as the tracked version. We need
  for t in tqdm(ds_hist["time"]):
    #bool = ds_hist["TM90"].loc[t,:,plat,plon] > 0
    zg += ds_hist["zg"].loc[t].values[:,:]
    #to convert it to boolean values
    bool = sum(sum(TM90.loc[t,plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon])) > 0.1*(l_lon*2/2.5)*(l_lat*2/2.5) #10 percent condition
    if bool == True:
      N_h += 1
      zg_h += ds_hist["zg"].loc[t].values[:,:]
    if bool == False:
      N_z += 1
      zg_z += ds_hist["zg"].loc[t].values[:,:]
  f_h = N_h/(N_h + N_z)
  f_z = 1-f_h
  # we have to adjust zg dimensions
  zg = zg/((N_b+N_z)*g)
  zg_b = zg_b/(N_b*g)
  zg_z = zg_z/(N_z*g)
  
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/f_b_"+version+"_"+region,f_b)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/f_z_"+version+"_"+region,f_z)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_"+version+"_"+region,zg)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_b_"+version+"_"+region,zg_b)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_z_"+version+"_"+region,zg_z)

  #PRESENT
  version = "pres"
  N_b = 0
  N_z = 0
  zg = np.zeros(np.shape(ds_pres["zg"].values[0,:,:]))
  zg_b = np.zeros(np.shape(ds_pres["zg"].values[0,:,:]))
  zg_z = np.zeros(np.shape(ds_pres["zg"].values[0,:,:]))
  TM90 = ds_pres["DAV"]
  TM90 = TM90[:,:,:]
  TM90 = xr.where(TM90.loc[:,:,:] > 0,1,0) #don't know why but it was just the same as the tracked version. We need
  for t in tqdm(ds_pres["time"]):
    #bool = ds_pres["TM90"].loc[t,:,plat,plon] > 0
    zg += ds_pres["zg"].loc[t].values[:,:]
    #to convert it to boolean values
    bool = sum(sum(TM90.loc[t,plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon])) > 0.1*(l_lon*2/2.5)*(l_lat*2/2.5) #10 percent condition
    if bool == True:
      N_b += 1
      zg_b += ds_pres["zg"].loc[t].values[:,:]
    if bool == False:
      N_z += 1
      zg_z += ds_pres["zg"].loc[t].values[:,:]
  f_b = N_b/(N_b + N_z)
  f_z = 1-f_b
  # we have to adjust zg dimensions
  zg = zg/((N_b+N_z)*g)
  zg_b = zg_b/(N_b*g)
  zg_z = zg_z/(N_z*g)
  
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/f_b_"+version+"_"+region,f_b)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/f_z_"+version+"_"+region,f_z)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_"+version+"_"+region,zg)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_b_"+version+"_"+region,zg_b)
  np.save("/home/michele/prog/summer_greenland_block_prel/decomposition/data/zg_z_"+version+"_"+region,zg_z)
```

```python
#import data

version = "hist"
f_b_h = np.load("data/f_b_"+version+"_"+region+".npy", allow_pickle = True)
f_z_h = np.load("data/f_z_"+version+"_"+region+".npy", allow_pickle = True)
zg_h = np.load("data/zg_"+version+"_"+region+".npy", allow_pickle = True)
zg_b_h = np.load("data/zg_b_"+version+"_"+region+".npy", allow_pickle = True)
zg_z_h = np.load("data/zg_z_"+version+"_"+region+".npy", allow_pickle = True)

version = "pres"
f_b_p = np.load("data/f_b_"+version+"_"+region+".npy", allow_pickle = True)
f_z_p = np.load("data/f_z_"+version+"_"+region+".npy", allow_pickle = True)
zg_p = np.load("data/zg_"+version+"_"+region+".npy", allow_pickle = True)
zg_b_p = np.load("data/zg_b_"+version+"_"+region+".npy", allow_pickle = True)
zg_z_p = np.load("data/zg_z_"+version+"_"+region+".npy", allow_pickle = True)

#lon = np.load("data/lon"+".npy")
#lat = np.load("data/lat"+".npy") 

print(f_b_h,f_b_p)
```

```python
#plot

fig = plt.figure(figsize=(15,10))



box_vertices = [[plon-l_lon, plat-l_lat], [plon+l_lon, plat-l_lat], 
                [plon+l_lon, plat+l_lat], [plon-l_lon, plat+l_lat], [plon-l_lon, plat-l_lat]]


#1
dataarray_comp = xr.DataArray(data=zg_p-zg_h,dims=["lat","lon"],coords = dict(lat=lat,lon=lon))
RMS = sum(sum(dataarray_comp.loc[plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon]**2))
RMS = float(RMS)
dataarray_comp.name = "zg_an"
ds_1 = dataarray_comp.to_dataset()


ax1 = BP.PlotField(ds = ds_1,
              field="zg_an", #name of zg500 matrix, may vary from file to file
              subplot_grid = 22,
              subplot_idx = 1,
              box_vertices = box_vertices,
              extent =[-120,60,20,90],
              plot_title="(a) TOTAL DIFFERENCE; RMSD: " + "{:.2f}".format(RMS) + " m^2")

#2
dataarray_comp = xr.DataArray(data=(-f_b_h+f_b_p)*(+zg_b_p-zg_z_p),dims=["lat","lon"],coords = dict(lat=lat,lon=lon))
RMS = sum(sum(dataarray_comp.loc[plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon]**2))
RMS = float(RMS)
dataarray_comp.name = "zg_an"
ds_2 = dataarray_comp.to_dataset()

ax2 = BP.PlotField(ds = ds_2,
              field="zg_an", #name of zg500 matrix, may vary from file to file
              subplot_grid = 22,
              subplot_idx = 2,
              box_vertices = box_vertices,
              extent =[-120,60,20,90],
              plot_title="(b) FREQUENCY DIFFERENCE; RMSD: " + "{:.2f}".format(RMS)+ " m^2")

#3
dataarray_comp = xr.DataArray(data=(f_b_h)*(-(zg_b_h-zg_b_p)+(zg_z_h-zg_z_p)),dims=["lat","lon"],coords = dict(lat=lat,lon=lon))
RMS = sum(sum(dataarray_comp.loc[plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon]**2))
RMS = float(RMS)
dataarray_comp.name = "zg_an"
ds_3 = dataarray_comp.to_dataset()

ax3 = BP.PlotField(ds = ds_3,
              field="zg_an", #name of zg500 matrix, may vary from file to file
              subplot_grid = 22,
              subplot_idx = 3,
              box_vertices = box_vertices,
              extent =[-120,60,20,90],
              plot_title="(c) BLOCKING PATTERN DIFFERENCE; RMSD: " + "{:.2f}".format(RMS)+ " m^2")

#4
dataarray_comp = xr.DataArray(data=-zg_z_h+zg_z_p,dims=["lat","lon"],coords = dict(lat=lat,lon=lon))
RMS = sum(sum(dataarray_comp.loc[plat-l_lat:plat+l_lat,plon-l_lon:plon+l_lon]**2))
RMS = float(RMS)
dataarray_comp.name = "zg_an"
ds_4 = dataarray_comp.to_dataset()

ax4 = BP.PlotField(ds = ds_4,
              field="zg_an", #name of zg500 matrix, may vary from file to file
              subplot_grid = 22,
              subplot_idx = 4,
              box_vertices = box_vertices,
              extent =[-120,60,20,90],
              plot_title="(d) MEAN STATE DIFFERENCE; RMSD: " + "{:.2f}".format(RMS)+ " m^2")

plt.savefig(output_prefix + region + "_mas-mab_comparison.pdf",bbox_inches='tight',dpi=250)
```
